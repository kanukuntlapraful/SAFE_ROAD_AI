```python
import pandas as pd
import plotly.express as px

CSV_PATH = r"C:\Users\jashw\OneDrive\Documents\2-1\DAV\DAV Project\cleaned data\indian_roads_dataset_cleaned.csv"
df = pd.read_csv(CSV_PATH)

state_counts = (
    df.groupby("state")
      .size()
      .reset_index(name="Accidents")
      .sort_values("Accidents", ascending=False)
)

fig = px.bar(
    state_counts,
    x="state",
    y="Accidents",
    color="Accidents",
    color_continuous_scale="Reds",
    title="Road Accidents by State",
    labels={"state": "State", "Accidents": "Number of Accidents"},
)
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```

```python
# 2. Pie chart: accident severity distribution
severity_counts = df['accident_severity'].value_counts().reset_index()
severity_counts.columns = ['Severity', 'Accidents']
fig = px.pie(severity_counts, names='Severity', values='Accidents', title='Distribution of Accident Severity', hole=0.35)
fig.show()
```

```python
# 3. Histogram: risk score distribution
fig = px.histogram(df, x='risk_score', nbins=20, color='accident_severity', title='Distribution of Accident Risk Scores', labels={'risk_score':'Risk Score','count':'Number of Accidents'})
fig.show()
```

```python
# 4. Scatter plot: risk score vs casualties
fig = px.scatter(df, x='risk_score', y='casualties', color='accident_severity', size='vehicles_involved', hover_name='city', hover_data=['state','road_type','cause'], title='Accident Risk Score vs Casualties', labels={'risk_score':'Risk Score','casualties':'Casualties'})
fig.show()
```

```python
# 5. Heatmap: correlations between numeric accident measures
numeric_cols = ['hour','lanes','vehicles_involved','casualties','risk_score','temperature']
corr = df[numeric_cols].corr(numeric_only=True)
fig = px.imshow(corr, text_auto=True, color_continuous_scale='RdBu_r', zmin=-1, zmax=1, title='Correlation Heatmap of Accident Variables')
fig.show()
```

```python
# 6. Geographic plot: accident locations in India
fig = px.scatter_geo(df, lat='latitude', lon='longitude', color='accident_severity', size='risk_score', hover_name='city', hover_data=['state','road_type','cause','casualties'], scope='asia', projection='natural earth', title='Road Accident Locations in India')
fig.update_geos(lataxis_range=[6,38], lonaxis_range=[68,98], showcountries=True)
fig.show()
```

```python
# 7. Bar chart: accidents by hour of day
hour_counts = df.groupby('hour').size().reset_index(name='Accidents')
fig = px.bar(hour_counts, x='hour', y='Accidents', title='Road Accidents by Hour of Day', labels={'hour':'Hour (0–23)','Accidents':'Number of Accidents'})
fig.update_xaxes(dtick=1)
fig.show()
```
